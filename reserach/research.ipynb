# Part 7 : Data Preparation for Machine Learning

## 7.1 Introduction

Before training machine learning models, the dataset must be converted into a format suitable for learning algorithms.

This stage prepares the dataset by separating the input features from the target variable, encoding categorical variables, splitting the dataset into training and testing sets, and scaling numerical features when required.

These preprocessing steps ensure that the machine learning models receive clean, consistent, and standardized data for effective training.

## 7.2 Create Feature Matrix (X) and Target Variable (y)

Machine learning models require two components:

- **Feature Matrix (X):** Contains all independent variables used for prediction.
- **Target Variable (y):** Contains the output variable that the model learns to predict.

````python
X = df_preprocessed.copy()

# Encode categorical columns
categorical_features = [
    "bp_category",
    "bp_category_encoded"
]

for feature in categorical_features:
    X[feature], _ = pd.factorize(X[feature].astype(str))

# Keep only numerical columns
X = X.select_dtypes(include=[np.number]).drop(columns=["cardio"])

# Target Variable
y = df_preprocessed["cardio"]

print("Feature Matrix Shape :", X.shape)
print("Target Variable Shape :", y.shape)
````

### Observation

The dataset has been successfully divided into:

- Feature Matrix (X)
- Target Variable (y)

These datasets will be used for training and evaluating machine learning models.

## 7.3 Train-Test Split

To evaluate the model fairly, the dataset is divided into training and testing subsets.

- **Training Set:** Used to train the machine learning model.
- **Testing Set:** Used to evaluate the performance of the trained model on unseen data.

In this project, an **80:20 split** is used.

````python
from sklearn.model_selection import train_test_split
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
````

````python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
````

### Observation

The dataset has been divided into:

- 80% Training Data
- 20% Testing Data

Stratified sampling ensures that both datasets maintain a similar class distribution.

## 7.4 Feature Scaling

Many machine learning algorithms perform better when numerical features are on a similar scale.

Standardization transforms numerical features so that they have:

- Mean = 0
- Standard Deviation = 1

The scaler is fitted only on the training data and then applied to both training and testing datasets to prevent data leakage.

````python
from sklearn.preprocessing import StandardScaler
````

````python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
````

````python
print("Training Data Shape :", X_train_scaled.shape)
print("Testing Data Shape  :", X_test_scaled.shape)
````

### Observation

The numerical features have been standardized successfully. The transformed datasets are now ready for machine learning model development.

# Part 8 : Model Development

## 8.1 Introduction

Model Development is the process of training machine learning algorithms using the prepared dataset. The objective is to identify the model that best predicts the presence of heart disease.

In this section, multiple classification algorithms will be trained using the same training dataset. Their performance will be evaluated and compared using standard evaluation metrics in the next section.

The algorithms selected include:

- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)
- Naïve Bayes

## 8.2 Import Machine Learning Algorithms

````python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
````

## 8.3 Initialize Machine Learning Models

Initialize all machine learning models with their default parameters. A fixed random state is used wherever applicable to ensure reproducibility.

````python
models = {

    "Logistic Regression": LogisticRegression(random_state=42),

    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "Random Forest": RandomForestClassifier(random_state=42),

    "Support Vector Machine": SVC(random_state=42),

    "K-Nearest Neighbors": KNeighborsClassifier(),

    "Naive Bayes": GaussianNB(),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss"
    )

}
````

## 8.4 Train Machine Learning Models

Train each machine learning algorithm using the training dataset.

````python
trained_models = {}

for model_name, model in models.items():

    model.fit(X_train_scaled, y_train)

    trained_models[model_name] = model

print("All models trained successfully.")
````

## 8.5 Generate Predictions

Generate predictions for the testing dataset using each trained model.

````python
predictions = {}

for model_name, model in trained_models.items():

    predictions[model_name] = model.predict(X_test_scaled)

print("Predictions generated successfully.")
````

## 8.6 Verify Trained Models

````python
trained_models.keys()
````

# Part 9 : Model Evaluation

## 9.1 Introduction

Model Evaluation is the process of assessing the performance of trained machine learning models using various evaluation metrics.

The objective of this section is to compare the performance of all trained models and identify the most suitable model for heart disease prediction.

The following evaluation metrics will be used:

- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC Score

## 9.2 Import Evaluation Metrics

````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
````

## 9.3 Evaluate All Models

Evaluate every trained model using the testing dataset and calculate the performance metrics.

````python
evaluation_results = []

for model_name, model in trained_models.items():

    y_pred = model.predict(X_test_scaled)

    evaluation_results.append({

        "Model": model_name,

        "Accuracy": accuracy_score(y_test, y_pred),

        "Precision": precision_score(y_test, y_pred),

        "Recall": recall_score(y_test, y_pred),

        "F1-Score": f1_score(y_test, y_pred),

        "ROC-AUC": roc_auc_score(y_test, y_pred)

    })

evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df
````

## 9.4 Model Performance Comparison

````python
evaluation_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)
````

### Observation

The comparison table summarizes the performance of all machine learning models based on multiple evaluation metrics.

The model with consistently higher scores across these metrics will be selected for further optimization.

## 9.5 Visual Comparison of Model Performance

````python
plt.figure(figsize=(10,5))

sns.barplot(
    data=evaluation_df.sort_values("Accuracy", ascending=False),
    x="Accuracy",
    y="Model"
)

plt.title("Model Accuracy Comparison")

plt.xlabel("Accuracy")

plt.ylabel("Machine Learning Models")

plt.show()
````

## 9.6 Identify the Best Model

````python
best_model = evaluation_df.sort_values(
    by="Accuracy",
    ascending=False
).iloc[0]

best_model
````

````python
print(f"Best Model : {best_model['Model']}")
print(f"Accuracy   : {best_model['Accuracy']:.4f}")
````

### Observation

The model with the highest evaluation score is selected as the candidate for hyperparameter tuning in the next phase.

## 9.7 Classification Report

Generate a detailed classification report for each trained model.

````python
for model_name, model in trained_models.items():

    print("=" * 70)

    print(f"{model_name}")

    print("=" * 70)

    y_pred = model.predict(X_test_scaled)

    print(classification_report(y_test, y_pred))
````

## 9.8 Confusion Matrix

Visualize the confusion matrix for the best-performing model.

````python
best_model_name = best_model["Model"]

best_classifier = trained_models[best_model_name]

best_prediction = best_classifier.predict(X_test_scaled)

cm = confusion_matrix(y_test, best_prediction)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(f"Confusion Matrix ({best_model_name})")

plt.xlabel("Predicted Label")

plt.ylabel("Actual Label")

plt.show()
````

### Summary
- Purpose: Compare and assess all trained models to select the best candidate for tuning.
- Metrics: Accuracy, Precision, Recall, F1-Score, ROC-AUC (plus confusion matrix and classification report).
- Process: Generate predictions on test set, compute metrics, collect into evaluation_df, and sort by performance.
- Visualization: Barplot comparing model accuracies for quick visual ranking.
- Selection: Pick the top model (highest accuracy) as the candidate for hyperparameter tuning.
- Diagnostics: Print classification reports for every model and plot the confusion matrix for the chosen best model.

# Part 10 : Cross-Validation Analysis

## 10.1 Introduction

Cross-validation is a model validation technique used to evaluate the generalization performance of a machine learning model. Instead of relying on a single train-test split, the training dataset is divided into multiple folds.

The model is trained on different subsets of the data and evaluated repeatedly. This provides a more reliable estimate of model performance and helps detect overfitting or underfitting.

In this project, **5-Fold Cross-Validation** is used.

## 10.2 Import Cross-Validation Module

# Part 11 : Hyperparameter Tuning

## 11.1 Introduction

Hyperparameter tuning is the process of finding the optimal combination of hyperparameters that improves the predictive performance of a machine learning model.

Based on the model evaluation and cross-validation results, the **Support Vector Machine (SVM)** achieved the best overall performance and was selected for optimization.

### Computational Challenge

The training dataset contains **54,564 samples**, making exhaustive hyperparameter tuning computationally expensive.

Initially, **GridSearchCV** was considered for hyperparameter optimization. However, evaluating every possible parameter combination required a large number of model fits, resulting in a very long execution time on the available hardware.

To overcome this challenge while maintaining reliable model performance, **RandomizedSearchCV** was adopted. Furthermore, hyperparameter tuning was performed on a representative subset of the training data. After identifying the optimal hyperparameters, the final SVM model was retrained using the complete training dataset.

This strategy significantly reduces computational cost while maintaining the quality of the final model.

## 11.2 Import GridSearchCV

GridSearchCV performs an exhaustive search over a predefined set of hyperparameter values using cross-validation.

````python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.model_selection import train_test_split
````

## 11.3 Create a Representative Tuning Dataset

Instead of performing hyperparameter tuning on the entire training dataset, a representative subset is created.

The subset preserves the original class distribution using stratified sampling and substantially reduces computational time during the tuning process.

````python
X_tune, _, y_tune, _ = train_test_split(
    X_train_scaled,
    y_train,
    train_size=10000,
    stratify=y_train,
    random_state=42
)

print("Tuning Dataset Shape :", X_tune.shape)
````

### Observation

The tuning dataset contains a representative sample of the original training data.

This subset is used only for hyperparameter optimization. The final optimized model will still be trained using the complete training dataset.

## 11.4 Define the Hyperparameter Search Space

The following hyperparameters are selected for optimization.

To reduce computational complexity, only the most commonly used SVM kernel (**RBF**) is considered because it achieved the best performance during model evaluation.

````python
param_distributions = {

    "C":[
        0.1,
        1,
        10,
        100
    ],

    "gamma":[
        "scale",
        "auto"
    ],

    "kernel":[
        "rbf"
    ]

}
````

## 11.4 Initialize the SVM Model

````python
svm_model = SVC(random_state=42)
````

## 11.6 Perform Hyperparameter Tuning

RandomizedSearchCV randomly samples combinations from the predefined search space.

Compared to GridSearchCV, this approach significantly reduces execution time while still identifying high-quality hyperparameter combinations.

````python
random_search = RandomizedSearchCV(

    estimator=svm_model,

    param_distributions=param_distributions,

    n_iter=4,

    cv=5,

    scoring="accuracy",

    random_state=42,

    n_jobs=-1,

    verbose=2

)

random_search.fit(
    X_tune,
    y_tune
)
````

## 11.7 Best Hyperparameters

````python
print("Best Hyperparameters")

random_search.best_params_
````

## 11.8 Best Cross-Validation Score

````python
print(
    f"Best Cross Validation Accuracy : {random_search.best_score_:.4f}"
)
````

## 11.9 Train the Optimized Model on the Complete Training Dataset

After identifying the best hyperparameters, the optimized SVM model is retrained using the complete training dataset.

````python
best_svm = random_search.best_estimator_

best_svm.fit(
    X_train_scaled,
    y_train
)
````

## 11.10 Evaluate the Optimized Model

````python
optimized_prediction = best_svm.predict(
    X_test_scaled
)

optimized_accuracy = accuracy_score(
    y_test,
    optimized_prediction
)

optimized_precision = precision_score(
    y_test,
    optimized_prediction
)

optimized_recall = recall_score(
    y_test,
    optimized_prediction
)

optimized_f1 = f1_score(
    y_test,
    optimized_prediction
)

optimized_auc = roc_auc_score(
    y_test,
    optimized_prediction
)
````

````python
print(f"Accuracy  : {optimized_accuracy:.4f}")
print(f"Precision : {optimized_precision:.4f}")
print(f"Recall    : {optimized_recall:.4f}")
print(f"F1 Score  : {optimized_f1:.4f}")
print(f"ROC-AUC   : {optimized_auc:.4f}")
````

### Summary:-
- Goal: Optimize the SVM chosen from cross-validation to improve predictive performance.
- Compute challenge: Full GridSearchCV is prohibitively expensive on ~54k training samples.
- Strategy: Use RandomizedSearchCV on a representative stratified subset (10,000 samples) to find good hyperparameters, then retrain the best estimator on the complete training set.
- Search space: `C = [0.1, 1, 10, 100]`; `gamma = ["scale","auto"]`; `kernel = ["rbf"]`.
- RandomizedSearchCV settings: `n_iter=4, cv=5, scoring="accuracy", random_state=42, n_jobs=-1, verbose=2`.
- Workflow: create `X_tune/y_tune` via stratified sampling → run `random_search.fit(X_tune,y_tune)` → inspect `best_params_` and `best_score_` → get `best_estimator_` and retrain on full `X_train_scaled/y_train` → evaluate on X_test_scaled (accuracy, precision, recall, F1, ROC-AUC) → compare original vs optimized SVM and plot results.
- Recommendation: If tuning remains slow, reduce `n_iter`, use fewer CV folds, use `HalvingGridSearchCV` / `RandomizedSearchCV` with a smaller subset, or perform tuning on a smaller feature/sample sample before finalizing.
